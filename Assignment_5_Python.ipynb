#### 1. Write a Python Program to Find Factorial of Number Using Recursion?

```python
def factorial(n):
    if n == 1:
        return 1
    return n*factorial(n-1)
print(factorial(5))
print(factorial(3))
print(factorial(6))
```

#### 2. Write a Python Program to calculate your Body Mass Index?

```python
def Body_Mass_Index(h,w):
    bmi = w/(h**2)
    if (bmi < 18.5):
        status = 'Underweight'
    elif (bmi >= 18.5 and bmi < 24.9):
        status = 'Healthy'
    elif (bmi >= 24.9 and bmi < 30):
        status = 'Overweight'
    elif (bmi >=30):
        status = 'Suffering from Obesity'
    print(f'Your\'re BMI is {bmi} and status is {status} ')

print("height in meters(m) and weight in kilograms(kg)")
Body_Mass_Index(1.7,70) # height in meters(m) and weight in kilograms(kg)
Body_Mass_Index(1.7,50)
```

#### 3. Write a Python Program to calculate the natural logarithm of any number?

```python
import math
def natural_log_n(n):
    print(math.log(n))
natural_log_n(49)
natural_log_n(50)
```

#### 4. Write a Python Program for cube sum of first n natural numbers?

```python
def cube_sum_n(n):
    return pow((n*(n+1)//2),2)
print(cube_sum_n(10))
print(cube_sum_n(5))
```
