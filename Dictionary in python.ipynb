```python
# Dictionary
d = {1:'a','A':2,6:4,(1,2):4, 5:[1,2,(3,4),5]}
print(d)
print(d.keys())
print(d.values())
print(d['A'])
```

```python
# empty dictionary
d = {}
d1 = dict()
# Dictionary with Values
d2 = {1:'a','A':2,6:4,(1,2):4, 5:[1,2,(3,4),5]}
```

```python
# Dictionary values access by using keys
d = {1:'a','A':2,6:4,(1,2):4, 5:[1,2,(3,4),5]}
print(d[1])
print(d['A'])
print(d[5])
```

```python
# dictionary with duplicate keys
# lets see what happens if duplicate key is created and then the dictionary is printed.
d = {1:'a','A':2,6:4,(1,2):4, 5:[1,2,(3,4),5],'A':10}
print(d)
# observe the output key A's value is updated to the new value. it's not an error
```

```python
d1 = {'language':'python','version':'3.13.5','level':'High level'}
d_keys = d1.keys()
d_values = d1.values()
d_items = d1.items()
print(type(d_keys),type(d_values),type(d_items))
print(d_keys)
print(d_values)
print(d_items)
```

```python
d1 = {'language':'python','version':'3.13.5','level':'High level'}
d_keys = d1.keys()
keys_list = list(d_keys)
print(keys_list)
print(type(keys_list))
d_values = d1.values()
values_list = list(d_values)
print(values_list)
print(type(values_list))
d_items = d1.items()
items_list = list(d_items)
print(items_list)
print(type(items_list))
```

```python
d2 = {'language':'python','version':'3.13.5','level':'High level',1:1,'b':2}
print(d2)
pop_key = d2.pop(1)
print(pop_key)
print(d2)
pop_item = d2.popitem()
print(pop_item)
print(d2)
```

```python
# Accessing and Updating the Values
d3 = {'language':'python','version':'3.13.5','level':'High level'}
print(d3)
print(d3['language'])
d3['A'] = 1
d3['language'] = 'Java'
print(d3)
```

```python
# Accessing and Updating using Dictionary Methods
# get(key,default)
d = {'language':'python','version':'3.13.5','level':'High level'}
val = d.get('p','hello')
print(d.get('level'))
print(val)
```

```python
# update(dictionary)
d = {'language':'python','version':'3.13.5'}
d1 = {'language':'java','level':'High level'}
d.update(d1)
print(d)
print(d1)
```

```python
d = {'language':'python','version':'3.13.5'}
d1 = {'language':'java','level':'High level'}
d1.update(d)
print(d1)
print(d)
```

```python
# clear() - Removes the all items from dictionary
d1 = {'language':'python','version':'3.13.5'}
d1.clear()
print(d1)
```
