```python
# set
s = {1,2,30,'a','b',34.7,'python'}
print(s)
```

```python
# Creating empty set
s1 = {}
s2 = set() # only way to create empty set
print(type(s1),type(s2))
```

```python
s4 = {1,2,30,'a','b',34.7,'python',(1,2,3,4)}
print(s4)
# print(s4[2]) --> TypeError: 'set' object is not subscriptable
```

```python
# read set of integers from user
s = set(map(int,input().split()))
print(s)
```

```python
# read set of strings from user
s = set(map(str,input().split()))
print(s)
```

```python
# set methods
s = {1,2,3,4}
print("Original set: ",s)
s.add(100)
print("After add operation:",s)
s.update([1,20,34])
print("After Update Operation:",s)
s.remove(2)
print("After removing 2 from set:",s)
s.discard(20)
print("After discarding 20 from set:",s)
val = s.pop()
print("After pop Operation:",val,s)
s.clear()
print("After clear Operation:",s)
```

```python
# union(), intersection(), difference(), symmetric_difference()
# if 3 sets are there then we use s1.union(s2,s3) as the syntax.
s1 = {1,2,3,4}
s2 = {3,4,5,6}
print("s1: ",s1)
print("s2: ",s2)
union = s1.union(s2)
print("Union of s1 and s2: ",union)
intersection = s1.intersection(s2)
print("intersection of s1 and s2: ",intersection)
difference = s1.difference(s2)
print("difference of s1 and s2: ",difference)
symmetric_difference = s1.symmetric_difference(s2)
print("symmetric_difference of s1 and s2: ",symmetric_difference)
```

```python
# issubset(), issuperset(), isdisjoint()
s1 = {1,2,3,4,5,6}
s2 = {4,6}
print(s1.issubset(s2))
print(s2.issubset(s1))
print(s1.issuperset(s2))
print(s1.isdisjoint(s2))
```
