#### 1. Write a Python Program to Find LCM?

```python
def lcm(*args):
    lst = list(args)
    max_num = max(lst)
    lcm_val = max_num

    while True:
        if all(lcm_val%n == 0 for n in lst):
            return lcm_val
        lcm_val += max_num
        
print(lcm(4,5,6))
print(lcm(3,7,14))
```

#### 2. Write a Python Program to Find HCF?

```python
def hcf(*args):
    lst = list(args)
    min_num = min(lst)

    for i in range(min_num,0,-1):
        if all(n%i == 0 for n in lst):
            return i
print(hcf(12, 18, 24)) 
print(hcf(5, 10, 15))  
print(hcf(7, 13, 19))
```

#### 3. Write a Python Program to Convert Decimal to Binary, Octal and Hexadecimal?

```python
n = int(input())
print(f"Binary number of the {n}:",bin(n))
print(f"Octal number of the {n}:",oct(n))
print(f"Hexadecimal number of the {n}:",hex(n))
```

#### 4. Write a Python Program To Find ASCII value of a character?

```python
char = input()
print(f"ASCII value of the character '{char}':",ord(char))
```

#### 5. Write a Python Program to Display Fibonacci Sequence Using Recursion?

```python
def fibonacci(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    return fibonacci(n-1)+fibonacci(n-2)
n = int(input())
for i in range(n):
    print(fibonacci(i),end=" ")
```
