#### 1. Library Management System (Encapsulation + Composition) Create classes Book and Library.Each book has attributes: title, author, b_no, and   available_copies. Library should manage a collection of books with methods:  
add_book(book)  
borrow_book(b_no) (decrease available copies if available)  
return_book(b_no) (increase copies back)  
Demonstrate by adding books and borrowing them.  

```python
class Book:
    def __init__(self,title,author,b_no,avail_copies):
        self.__title = title
        self.__author = author
        self.__b_no = b_no
        self.__avail_copies = avail_copies

    def get_bno(self):
        return self.__b_no

    def get_title(self):
        return self.__title

    def get_author(self):
        return self.__author

    def get_avail_copies(self):
        return self.__avail_copies

    def borrow_book(self):
        if self.__avail_copies > 0:
            self.__avail_copies -= 1
            return True
        return False

    def return_book(self):
        self.__avail_copies += 1

    def __str__(self):
        return f"[{self.__b_no}] {self.__title} by {self.__author} (Available: {self.__avail_copies})"

class Library:
    def __init__(self):
        self.books = {}
        
    def add_book(self,book):
        self.books[book.get_bno()] = book
        print(f"Book '{book.get_title()}' added.")
        
    def borrow_book(self,b_no):
        if b_no in self.books and self.books[b_no].borrow_book():
            print(f"Book {b_no} borrowed successfully")
        else:
            print(f"Book {b_no} not available")

    def return_book(self,b_no):
        if b_no in self.books:
            self.books[b_no].return_book()
            print(f"Book {b_no} returned successfully")
        else:
            print(f"Book {b_no} not found in the library")

    def display_books(self):
        if not self.books:
            print("Library has no books.")
        else:
            print("\n Library Collections:")
            for book in self.books.values():
                print(book)

l = Library()
b1 = Book('Harry Potter','J.K Rowling','B134',7)
b2 = Book('Dark Wells','Bishop Son','B012',3)
l.add_book(b1)
l.add_book(b2)
l.display_books()
l.borrow_book('B134')
l.display_books()
l.borrow_book('B012')
l.borrow_book('B012')
l.borrow_book('B012')
l.borrow_book('B012')
l.display_books()
l.return_book('B012')
l.display_books()
```

#### 2. Employee Management (Polymorphism + Special Methods) Create a base class Employee with name, id, salary. Create two subclasses:  
Manager (extra attribute: bonus)  
Developer (extra attribute: programming_language)  
--> Implement \_\_str\_\_() to print details in a clean format.  
Write a function calculate_total_expense(employee_list) that calculates salaries + bonuses.  

```python
class Employee:
    def __init__(self,name,i_d,salary):
        self.name = name
        self.i_d = i_d
        self.salary = salary
        employee_list = {}

    def get_total_compensation(self):
        return self.salary
        
    def __str__(self):
        return f"Employee [ID: {self.name}, Name: {self.i_d}, Salary: {self.salary}]"
        
class Manager(Employee):
    def __init__(self,name,i_d,salary,bonus):
        super().__init__(name,i_d,salary)
        self.bonus = bonus

    def get_total_compensation(self):
        return self.salary + self.bonus

    def __str__(self):
        return f"Manager [ID: {self.i_d}, Name: {self.name}, Salary: {self.salary}, Bonus: {self.bonus}]"

class Developer(Employee):
    def __init__(self,name,i_d,salary,programming_language):
        super().__init__(name,i_d,salary)
        self.programming_language = programming_language

    def __str__(self):
        return f"Developer [ID: {self.i_d}, Name: {self.name}, Salary: {self.salary}, Programming Language: {self.programming_language}]"

def calculate_total_expense(employee_list):
    total = 0
    for emp in employee_list:
        total += emp.get_total_compensation()
    return total

if __name__ == '__main__':
    e1 = Manager('John','M100',50000,10000)
    e2 = Manager('Revi','M456',40000,5000)
    e3 = Developer('Ram','D456',60000,'Golang')
    e4 = Employee('Veena','E2134',50000)

    e = [e1,e2,e3,e4]
    
    for i in e:
        print(i)
    print(f"Total Company Expense: {calculate_total_expense(e)}")        
```

#### 3. Using hybrid inheritance develope a Advanced Calculator to perform basic and advanced level calculations.

```python
import math as m
import os
class basic_Calc:
    def __init__(self,a,b):
        self.a = a
        self.b = b

    def add(self):
        return sum([self.a,self.b])

    def sub(self):
        return self.b - self.a

    def mul(self):
        return self.a*self.b

    def div(self):
        return self.a/self.b
        
class Advanced_Calc(basic_Calc):
    def __init__(self,a,b):
        super().__init__(a,b)

    def power(self):
        return pow(self.a,self.b)

    def square_root(self):
        x = m.sqrt(self.a)
        y = m.sqrt(self.b)
        return f"SqureRoots of a and b are: {x, y}"

    def log_of(self):
        x = m.log(self.a)
        y = m.log(self.b)
        return f"log of a and b : {x, y}"

    def lcm_of(self):
        return m.lcm(self.a,self.b)

    def gcd_of(self):
        return m.gcd(self.a,self.b)

while True:
    print("Enter the operation number you want to perform:")
    print("1. Addition \n 2. Subtraction \n 3. Multiplication. \n 4. Division \n 5. power \n 6. Square Root \n 7. Logrithmic Value \n 8. LCM \n 9. GCD \n 10. Exit \n")
    o = int(input())
    if o == 1:
        print("Enter Values to perform operation")
        a,b = map(int,input().split())
        obj = Advanced_Calc(a,b)
        print(f"Addition of given values is : {obj.add()}")
    elif o == 2:
        print("Enter Values to perform operation")
        a,b = map(int,input().split())
        obj = Advanced_Calc(a,b)
        print(f"Subraction of given values is : {obj.sub()}")
    elif o == 3:
        print("Enter Values to perform operation")
        a,b = map(int,input().split())
        obj = Advanced_Calc(a,b)
        print(f"Multiplication of given values is : {obj.mul()}")
    elif o == 4:
        print("Enter Values to perform operation")
        a,b = map(int,input().split())
        obj = Advanced_Calc(a,b)
        print(f"Division of given values is : {obj.div()}")
    elif o == 5:
        print("Enter Values to perform operation")
        a,b = map(int,input().split())
        obj = Advanced_Calc(a,b)
        print(f"power of given values is : {obj.power()}")
    elif o == 6:
        print("Enter Values to perform operation")
        a,b = map(int,input().split())
        obj = Advanced_Calc(a,b)
        print(f"Square Root of given values is : {obj.square_root()}")
    elif o == 7:
        print("Enter Values to perform operation")
        a,b = map(int,input().split())
        obj = Advanced_Calc(a,b)
        print(f"Logarithm of given values is : {obj.log_of()}")
    elif o == 8:
        print("Enter Values to perform operation")
        a,b = map(int,input().split())
        obj = Advanced_Calc(a,b)
        print(f"LCM of given values is : {obj.lcm_of()}")
    elif o == 9:
        print("Enter Values to perform operation")
        a,b = map(int,input().split())
        obj = Advanced_Calc(a,b)
        print(f"GCD of given values is : {obj.gcd_of()}")
    else:
        print("-----END-----")
        break   
```
