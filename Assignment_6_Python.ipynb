#### 1. Write a Python Program to find largest element in an array?

```python
l1 = list(map(int,input().split()))
maxx = l1[0]
for i in range(1,len(l1)):
    if l1[i]>maxx:
        maxx = l1[i]
print(maxx)
```

#### 2. Write a Python Program to find smallest element in an array?

```python
lst = list(map(int,input().split()))
minn = lst[0]
for i in range(1,len(lst)):
    if lst[i]<minn:
        minn = lst[i]
print(minn)
```

#### 3. Write a Python Program to find second largest element in an array?

```python
lst = list(map(int,input().split()))
lst.sort()
print(lst[-2])
```

#### 5. Write a Python Program to check if given array is Monotonic?

```python
def checkMonotonic(arr):
    if (all(arr[i]<=arr[i+1] for i in range(len(arr)-1)) or all(arr[i]>=arr[i+1] for i in range(len(arr)-1))):
        return "Array is Monotonic"
    else:
        return "Array is not Monotonic"
print(checkMonotonic([1,2,3,4,5,6]))
print(checkMonotonic([4,5,1,2,6,9]))
print(checkMonotonic([1,2,3,3,4,5,6,6]))
print(checkMonotonic([7,6,4,4,3,1]))
```
