```python
# Check whether a number is positive, negative, or zero and even/odd.
n1 = int(input())
if n1<0:
    if n1%2 == 0:
        print(f"{n1} is negative even number.")
    else:
        print(f"{n1} is negative odd number.")
else:
    if n1 == 0:
        print(f"{n1} is a zero")
    else:
        if n1%2 == 0:
            print(f"{n1} is positive even number.")
        else:
            print(f"{n1} is positive odd number.")
```

```python
# Assign grades based on marks (90+: A+, 80-89: A, etc.).
m = int(input())
if 90<=m<=100:
    print("Grade is: A+")
elif 80<=m<=89:
    print("Grade is: A")
elif 70<=m<=79:
    print("Grade is: B")
elif 60<=m<=69:
    print("Grade is: C")
elif 35<=m<=59:
    print("Grade is: D")
else:
    print("Grade is: E")   
```

```python
# Create a simple calculator using +, -, *, / operators.
a ,b = map(int,input().split())
op = input()
if op == '+':
    c = a+b
    print(f"{a} + {b} = {c}")
elif op == '-':
    c = a-b
    print(f"{a} - {b} = {c}")
elif op == '*':
    c = a*b
    print(f"{a} * {b} = {c}")
else:
    c = a/b
    print(f"{a} / {b} = {c:.2f}")
```

```python
# Determine the type of triangle (Equilateral, Isosceles, Scalene) using sides.
s1, s2, s3 = map(int,input().split())
if (s1 == s2 and s2 == s3):
    print("It's an Equilateral Triangle.")
elif (s1 == s2 or s2 == s3 or s3 == s2):
    print("It's an Isosceles Triangle.")
else:
    print("It's a Scalene Triangle.")
     
```

```python
# Check if a day number (1–7) corresponds to weekday or weekend.
day_num = int(input("Enter a number between (1-7): "))
if day_num in (1,2,3,4,5):
    print("It's a Weekday")
elif day_num in (6,7):
    print("It's a Weekend")
else:
    print("Invalid day number. please enter a number between (1-7)")
    
```

```python
# Find the roots of a quadratic equation and display their nature.
a, b, c = map(int, input().split())
d = (b**2) - (4 * a * c)
val = (d) ** 0.5
root1 = (-b + val) / (2 * a)
root2 = (-b - val) / (2 * a)
if d == 0:
    print("The equation has two real equal roots")
elif d > 0:
    print("The equation has two distinct real roots")
else:
    print("The equation has two complex roots")
    
print("Roots are: ",root1,root2)
```
