#### 1. Create a function that takes a list of numbers between 1 and 10 (excluding one number) and returns the missing number.   
Testcases:   
missing_num([1, 2, 3, 4, 6, 7, 8, 9, 10]) ➞ 5   
missing_num([7, 2, 3, 6, 5, 9, 1, 4, 8]) ➞ 10   
missing_num([10, 5, 1, 2, 4, 6, 8, 3, 9]) ➞ 7   

```python
def missing_num(lst):
    missing_num = 0
    for i in range(1,11):
        if i in lst:
            continue
        else:
            missing_num = i
    return missing_num
print(missing_num([1, 2, 3, 4, 6, 7, 8, 9, 10]))
print(missing_num([7, 2, 3, 6, 5, 9, 1, 4, 8]))
print(missing_num([10, 5, 1, 2, 4, 6, 8, 3, 9]))
```

#### 2. Create a function that returns the thickness (in meters) of a piece of paper after folding it n number of times. The paper starts off with a thickness of 0.5mm.  
Testcases:  
num_layers(1) ➞ "0.001m" # Paper folded once is 1mm (equal to 0.001m)    
num_layers(4) ➞ "0.008m" # Paper folded 4 times is 8mm (equal to 0.008m)    
num_layers(21) ➞ "1048.576m" # Paper folded 21 times is 1048576mm (equal to 1048.576m)   

```python
def num_layers(n):
    thickness = 0.5 * (2 ** n)/1000
    return f"{thickness}m"
print(num_layers(1))
print(num_layers(4))
print(num_layers(21))
```

#### 3. Using list comprehensions, create a function that finds all even numbers from 1 to the given number.   
Testcases:   
find_even_nums(8) ➞ [2, 4, 6, 8]   
find_even_nums(4) ➞ [2, 4]   
find_even_nums(2) ➞ [2]  

```python
def find_even_nums(n):
    list_even = [x for x in range(1,n+1) if not x%2]
    return list_even
print(find_even_nums(8))
print(find_even_nums(4))
print(find_even_nums(2))
```

#### 4. Write a Python program to find the largest number in a list.   
Test Case 1  
Input: [4, 1, 7, 2]   
Output: 7   
Explanation: The largest number is 7.   
Test Case 2   
Input: [-5, -1, 0, 3]   
Output: 3   
Explanation: The largest number is 3.  

```python
def find_largest(lst):
    largest = 0
    for i in lst:
        if i>largest:
            largest = i
    return largest
print(find_largest([4, 1, 7, 2]))
print(find_largest([-5, -1, 0, 3]))           
```

#### 5. Write a Python program to find the second largest number in a list.   
Test Case 1   
Input: [10, 20, 4, 45, 99]   
Output: 45   
Explanation: Largest = 99, second largest = 45.   
Test Case 2   
Input: [5, 5, 10, 15]   
Output: 10   
Explanation: Unique sorted list = [5, 10, 15]. Second largest = 10.   

```python
def find_sec_largest(lst):
    req_lst = sorted(lst)
    return req_lst[-2]
print(find_sec_largest([10, 20, 4, 45, 99]))
print(find_sec_largest([5, 5, 10, 15]))
```

#### 6. Find Even Numbers in a String and Count Them Write a program that accepts a string containing digits and letters. Print all even digits found   in the string and the total count.    
Sample Input: Python Enter a string: a1b2c3d4   
Sample Output: Python Even numbers found: 2 4 Count of even numbers: 2   


```python
def find_digit_count(strr):
    req_str = ''
    c = 0
    for i in strr:
        if i.isdigit():
            if int(i)%2 == 0:
                req_str += i+' '
                c += 1
    req_str = req_str.strip()
    return f"Even numbers found: {req_str} Count of even numbers: {c}"
s = input("Python Enter a string: ")
res = find_digit_count(s)
print(res)       
```
