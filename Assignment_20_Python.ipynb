#### 2. Write a Python function that checks whether a given date falls on a weekend (Saturday or Sunday).Return "Yes" if it's a weekend, otherwise return "No".Input Format: A single line containing a date in the format:YYYY-MM-DD. Output Format:Print "Yes" if the date is a Saturday or Sunday, otherwise print "No".

```python
from datetime import date
import calendar
def is_weekend(d):
    day_num = d.weekday()
    day_name = calendar.day_name[day_num]
    if day_name == 'Saturday' or day_name == "Sunday":
        return True
    else:
        return False

d,m,y= map(int,input().split('-'))
d = date(y,m,d)
if is_weekend(d):
    print("Yes")
else:
    print("No")
```

#### 3. Given a time string in HH:MM 24-hour format, round the time to the nearest 15-minute interval.If the minutes are exactly halfway between two intervals (e.g., 7.5 minutes), round up. Return the rounded time in HH:MM format (24-hour clock). Note that rounding can cause the hour to roll over (e.g., 23:53 → 00:00).

```python
def return_rounded_time(strr):
    h,m = map(int,strr.split(":"))
    tot_mins = h*60 + m
    round_min = round(tot_mins/15)*15 
    if round_min == 24*60:
        round_min = 0
    round_hrs,round_mins = divmod(round_min,60)
    return f"{round_hrs:02d}:{round_mins:02d}"
s = input()
print(return_rounded_time(s))
```
