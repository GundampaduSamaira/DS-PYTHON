#### 1. Write a Python Program to Transpose a Matrix?

```python
def generate_transpose(in_matrix):
    out_matrix = []
    for ele in range(len(in_matrix[0])):
        out_matrix.append([0 for i in range(len(in_matrix))])
    for i in range(len(in_matrix)):
        for j in range(len(in_matrix[i])):
            out_matrix[j][i] = in_matrix[i][j]
    print(f'{in_matrix} -> {out_matrix}')

a = [[1,2,3],[4,5,6],[7,8,9]]
b = [[1,2],[4,5],[7,8]]
c = [[1,2,3],[4,5,6]]
        
generate_transpose(a)
generate_transpose(b)
generate_transpose(c)
```

#### 2. Write a Python Program to Sort Words in Alphabetic Order?

```python
def sort_Words(s):
    strr = s.title()
    lst = sorted(strr.split(" "))
    return ' '.join(lst)

print(sort_Words("Today is a Rainy Day"))
```

#### 3. Write a Python Program to Remove Punctuation From a String?

```python
def remove_punctuations(s):
    punctuations = '''()[]{}'"/\.,:;*&^%$#@!~`<>|'''
    res_str = ""
    for i in s:
        if i in punctuations:
            continue
        res_str += i
    return res_str
print(remove_punctuations("Hello *TH@is Y345e,'this' [yies] {}"))
print(remove_punctuations("ab}ghe*!a/<.>thisp\|pqari@%$@ren345l"))
```

#### 4. Write a Python program to check if the given number is a Disarium Number?

```python
def check_Disarium(num):
    strr = str(num)
    summ = 0
    k = 1
    for i in strr:
        summ += int(i)**k
        k += 1
    if summ == num:
        return f"{num} is a Disarium"
    else:
        return f"{num} is not a Disarium"
print(check_Disarium(89))
print(check_Disarium(135))
print(check_Disarium(45))
```

#### 5. Write a Python program to print all disarium numbers between 1 to 100?

```python
def check_Disarium(num):
    strr = str(num)
    summ = 0
    k = 1
    for i in strr:
        summ += int(i)**k
        k += 1
    if summ == num:
        return True
    else:
        return False
for i in range(1,101):
    if check_Disarium(i):
        print(i,end = " ")    
```
