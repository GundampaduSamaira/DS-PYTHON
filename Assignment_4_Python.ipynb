#### 1. Create a function that takes three parameters where:  
x is the start of the range (inclusive).  
y is the end of the range (inclusive).  
n is the divisor to be checked against.

Return an ordered list with numbers in the range that are divisible by the third parameter n. Return an empty list if there are no numbers that are  
divisible by n. Examples:  
list_operation(1, 10, 3) ➞ [3, 6, 9]  
list_operation(7, 9, 2) ➞ [8]  
list_operation(15, 20, 7) ➞ []  

```python
def divisible(a,b,c):
    s = set()
    for i in range(a,b+1):
        if i%c == 0:
            s.add(i)
    return s
a,b,c = map(int,input().split())
res = divisible(a,b,c)
print(res)           
```

#### 2. A group of friends have decided to start a secret society. The name will be the first letter of each of their names, sorted in a alphabetical order ? Create a function that takes in a list of names and returns the name of the secret society ?  

Examples:  
society_name(["Adam", "Sarah", "Malcolm"]) ➞ "AMS"  
society_name(["Harry", "Newt", "Luna", "Cho"]) ➞ "CHLN"  
society_name(["Phoebe", "Chandler", "Rachel", "Ross", "Monica", "Joey"]) ➞ "CJMPRR"

```python
def society_group(lst):
    lst.sort()
    res_str = ''
    for i in lst:
        res_str += i[0]
    return res_str

lst = []
n = int(input())
for i in range(n):
    e = input()
    if e == "":
        break
    else:
        lst.append(e)
res = society_group(lst)
print(res)
```

#### 4. Create a function that takes a string and returns True or False, depending on whether the characters are in order or not ?  
Examples:  
is_in_order("abc") ➞ True  
is_in_order("edabit") ➞ False  
is_in_order("123") ➞ True  
is_in_order("xyzz") ➞ True  

Notes:  
You don't have to handle empty strings.  

```python
def check_str_order(string):
    l = list(string)
    l.sort()
    req_str = ''.join(l)
    if string == req_str:
        return True
    else:
        return False
n = input()
res = check_str_order(n)
print(res)
```

#### 5. Create a function that takes a number as an argument and returns True or False depending on whether the number is symmetrical or not. A number  is symmetrical when it is the same as its reverse.  
Examples:  
is_symmetrical(7227) ➞ True  
is_symmetrical(12567) ➞ False  
is_symmetrical(44444444) ➞ True  
is_symmetrical(9939) ➞ False  
is_symmetrical(1112111) ➞ True  

```python
def is_symmetrical(strr):
    l = list(strr)
    l.reverse()
    req_str = ''.join(l)
    if req_str == strr:
        return True
    else:
        return False
n = input()
res = is_symmetrical(n)
print(res)  
```

#### 6. Given a string of numbers separated by a comma and space, return the product of the numbers.  
Examples:  
multiply_nums("2, 3") ➞ 6  
multiply_nums("1, 2, 3, 4") ➞ 24  
multiply_nums("54, 75, 453, 0") ➞ 0  
multiply_nums("10, -2") ➞ -20  

```python
def multiply_nums(strr):
    l = strr.split(',')
    p = 1
    for i in l:
        p *= int(i)
    return p

s = input()
res = multiply_nums(s)
print(res)
```
