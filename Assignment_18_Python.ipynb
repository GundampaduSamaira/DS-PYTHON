#### 1. Write a Python program to Create a base class Appliance with: An attribute power_rating set via input and A method show_spec() that prints:Power Rating: [power_rating] Create a derived class WashingMachine that: Adds an attribute load_capacity set via input Overrides show_spec() using super() and additionally prints: Load Capacity: [load_capacity]


```python
class Appliance:
    def __init__(self,power_rating):
        self.power_rating = power_rating
    def show_spec(self):
        return f"Power Rating: {self.power_rating}"
class WashingMachine(Appliance):
    def __init__(self,power_rating,load_capacity):
        super().__init__(power_rating)
        self.load_capacity = load_capacity
    def show_spec(self):
        print(super().show_spec())
        return f"Load Capacity: {self.load_capacity}"
pr = input()
lc = input()
wm = WashingMachine(pr,lc)
print(wm.show_spec())      
```

#### 2. Create a base class Math with a recursive method factorial(n) that computes the factorial of a number n.Create a derived class AdvancedMath that uses super().factorial(n) to get the factorial and prints:Factorial of [n] is [result]

```python
# first method
class Math:
    def factorial(self,n):
        if n == 0:
            return 1
        if n == 1:
            return 1
        return n*self.factorial(n-1)
class AdvancedMath(Math):
    def __init__(self,n):
        self.n = n
    def get_factorial(self):
        return f"Factorial of {self.n} is {super().factorial(self.n)}"
n = int(input())
am = AdvancedMath(n)
print(am.get_factorial())
```

#### 3. Write a python program to Create a class SchoolMember with a method details() that prints "School Member". Derive two classes: Teacher (inherits SchoolMember) with its own details() method that prints "Teacher of subject: [subject]". Student (inherits SchoolMember) with its own details() method that prints "Student of class: [grade]". Create a third class TeachingAssistant that inherits from both Teacher and Student.

```python
class SchoolMember:
    def details(self):
        print("SchoolMember")
        
class Teacher(SchoolMember):
    def __init__(self,subject):
        self.subject = subject
    def details(self):
        print(f"Teacher of Subject: {self.subject}")
        
class Student(SchoolMember):
    def __init__(self,grade):
        self.grade = grade
    def details(self):
        print(f"Student of Class: {self.grade}")
        
class TeachingAssistant(Teacher,Student):
    def __init__(self,subject,grade):
        Teacher.__init__(self,subject)
        Student.__init__(self,grade)
    def details(self):
        print(f"Teaching Assistant - Subject: {self.subject}, Class: {self.grade}")
        
m = SchoolMember()
m.details()

t = Teacher('Mathematics')
t.details()

s = Student('8th Grade')
s.details()

ta = TeachingAssistant('French','9th Grade')
ta.details()
```

#### 4. Create a base class Animal with a method sound() that prints "Animal makes a sound". Create two derived classes: Dog that overrides the sound() method to print "Bark" Cat that overrides the sound() method to print "Meow"

```python
class Animal:
    def sound(self):
        print("Animal makes a sound")
class Dog(Animal):
    def sound(self):
        print("Bark")
class Cat(Animal):
    def sound(self):
        print("Meow")
a = Animal()
d = Dog()
c = Cat()
a.sound()
d.sound()
c.sound()
```
