```python
# Write a python program for array rotation
# For example if array is [10,20,30,40,50]
# Then Array's one Left Rotation is--> [20,30,40,50,10]

n = int(input("Enter the length of array: "))
arr = list(map(int,input().split()))
temp = arr[0]
for i in range(n):
    if (i==(n-1)):
        arr[i] = temp
    else:
        arr[i] = arr[i+1]
print("Given array's one Left Rotation is:",arr)


# optimized simple code
# n = int(input("Enter the length of array: "))
# arr = list(map(int,input().split()))
# arr = arr[1:] + arr[:1]
# print(arr)
```

```python
# Write a python program for array rotation
# For example if array is [10,20,30,40,50]
# Then Array's one right Rotation --> [50,10,20,30,40]

n = int(input("Enter the length of array: "))
arr = list(map(int,input().split()))
temp = arr[-1]
for i in range(n-1,-1,-1):
    if i == 0:
        arr[i] = temp
    else:
        arr[i] = arr[i-1]
print("Given array's one right Rotation is:",arr)


# optimized simple code
# n = int(input("Enter the length of array: "))
# arr = list(map(int,input().split()))
# arr = arr[-1:] + arr[:n-1]
# print(arr)
```

```python
# Write a python program for k array rotations
# For example if array is [10,20,30,40,50]
# if k = 3 then array after 3 rotations is [40,50,10,20,30]
n = int(input("Enter the length of array: "))
arr = list(map(int,input().split()))
k = int(input())
k = n%k
arr = arr[k:] + arr[:k]
print(arr)
```

```python
# write a python program to split the array and add the first part to the end
# [10,20,30,40,50]
n = int(input("Enter the length of array: "))
arr = list(map(int,input().split()))
mid = n//2
arr = arr[mid:] + arr[:mid]
print(arr)
```
