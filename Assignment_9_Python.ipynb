#### 1. Write a Python program to check if the given number is Happy Number?

```python
def isHappyNum(n):
    seen = set()
    while True:
        summ = 0
        for i in str(n):
            summ += int(i)**2
        if summ == 1:
            return True
        if summ in seen:
            return False
        seen.add(summ)
        n = summ
    
n = int(input())
if isHappyNum(n):
    print(f"{n} is a Happy Number")
else:
    print(f"{n} is not a Happy Number")        
```

#### 2. Write a Python program to print all happy numbers between 1 and 100?

```python
def isHappyNum(n):
    seen = set()
    while True:
        summ = 0
        for i in str(n):
            summ += int(i)**2
        if summ == 1:
            return True
        if summ in seen:
            return False
        seen.add(summ)
        n = summ

print("Happy Numbers between 1 and 100 are: ")
for i in range(1,101):
    if isHappyNum(i):
        print(i,end = " ")
    
```

#### 3. Write a Python program to determine whether the given number is a Harshad Number?

```python
def isHarshadNum(num):
    s = str(num)
    summ = 0
    for i in s:
        summ += int(i)
    if num%summ == 0:
        return True
    else:
        return False
num = int(input())
if isHarshadNum(num):
    print(f"{num} is Harshad Number")
else:
    print(f"{num} is not a Harshad Number")
```

#### 5. Write a Python program to determine whether the given number is a Pronic Number?

```python
def isPronicNum(num):
    for i in range(1,int(pow(num,0.5))+1):
        if num == i*(i+1):
            return True
    else:
        return False
n = int(input())
if isPronicNum(n):
    print(f"{n} is a Pronic Number")
else:
    print(f"{n} is not a Pronic Number")
```
