#### 1. Write a Python to find all duplicate characters in a string?

```python
def find_dup(strr):
    freq_d = {}
    res = ''
    for i in strr.lower():
        if i not in freq_d:
            freq_d[i] = 1
        else:
            freq_d[i] += 1
    for i in freq_d:
        r = freq_d.get(i)
        if r > 1:
            res += i+','
    return f'Duplicate Characters are: {res}'

i = input()
r = find_dup(i)
print(r)

# Alternative Method
# def find_dup(strr):
#     non_dup = []
#     dup = []
#     for i in strr:
#         if i not in non_dup:
#             non_dup.append(i)
#         else:
#             dup.append(i)
#     return f'Duplicate Characters are:{list(set(dup))}'
# i = input()
# r = find_dup(i)
# print(r)
```

#### 2. Write a Python Program to check if a string contains any special character?

```python
def check_spl_char(s):
    spl_char = '[!@#$%^&*()_<>?/\|{}:;]'
    char_list = []
    c = 0
    for i in s:
        if i in spl_char:
            char_list.append(i)
            c += 1
    return f"There are {c} special characters in {s}: {char_list}"
            
i = input()
r = check_spl_char(i)
print(r)
```

#### 5. Write a Python program to Merge two Dictionaries?

```python
dict1 = {"apple": 1,"banana": 2,"cherry": 3,"date": 4,"mango": 5}
dict2 = {"car": 10,"bike": 20,"bus": 30,"train": 40,"plane": 50}
merged_dict = dict(list(dict1.items())+list(dict2.items()))
print(merged_dict)

# Alternative
dict1.update(dict2)
print(dict1)
```

#### 6. Write a Python program to convert key-values list to flat dictionary?

```python
pairs = [("ocean", 5),("forest", 12),("desert", 18),("island", 25),("volcano", 33),("robot", 41),("rocket", 56),("galaxy", 88),("universe", 100)]

print(dict(pairs))

# Alternative
req_dict = {}
for i in pairs:
    req_dict[i[0]] = i[1]
print(req_dict)
```

#### 9. Write a Python program to sort Python Dictionaries by Key or Value?

```python
d = {"zebra": 7,"apple": 3,"moon": 9,"river": 5,"castle": 12,"train": 1,"garden": 8,"ocean": 4}
def sort_dict(d,sort_type):
    if sort_type == 'key':
        print(dict(sorted(d.items(), key=lambda x:x[0])))
    else:
        print(dict(sorted(d.items(), key=lambda x:x[1])))
        
sort_dict(d,'key')        
sort_dict(d,'value')
```
