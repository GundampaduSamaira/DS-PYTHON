```python
# Write a program to check whether a number is positive or negative
n = int(input())
if n<0:
    print(f"{n} is Negative Number")
else:
    print(f"{n} is Positive Number")
```

```python
# Check whether a number is even or odd
n = int(input())
if n%2 == 0:
    print(f"{n} is Even Number")
else:
    print(f"{n} is Odd Number")
```

```python
# Find greater of two numbers
m = int(input())
n = int(input())
if m > n:
    print(f"{m} is Greater Number")
else:
    print(f"{n} is Greater Number")
```

```python
# Find the largest among three numbers
p = int(input())
q = int(input())
r = int(input())
if (p>q and q>r):
    print(f"{p} is largest number")
elif (q>r):
    print(f"{q} is larget number")
else:
    print(f"{r} is largest number")
    
    
```

```python
# Check if a number is divisible by 5 and 11.
n = int(input())
str_of_num = str(n)
odd_sum = 0
even_sum = 0
for i in range(len(str_of_num)):
    if (i%2 == 0):
        even_sum += int(str_of_num[i])
    else:
        odd_sum += int(str_of_num[i])
if even_sum > odd_sum:
    res = even_sum - odd_sum
else:
    res = odd_sum - even_sum
    
if (str_of_num[-1] == '5' or str_of_num[-1] == '0') and (res%11 == 0):
    print(f"{n} is divisible by both 5 and 11")
else:
    print(f"{n} is not divisible by both 5 and 11")
    
```

```python
# checking whether a year is leap year or not
year = int(input())
if (year % 4 == 0) and (year % 100 != 0 or year % 400 == 0):
    print(year,"is leap year")
else:
    print(year,"not a leap year")
```

```python
# Check whether a character is a vowel or a consonant
vowels = {'a','e','i','o','u'}
i = input()
if i.lower() in vowels:
    print(f"{i} is a Vowel")
else:
    print(f"{i} is a Consonant")
```

```python
# Check whether a number is a multiple of 3 or 7
n = int(input())
if (n%3 == 0 or n%7 == 0):
    print(f"{n} is multiple of 3 or 7")
else:
    print(f"{n} is not multiple of 3 or 7")
```

```python
# Check whether a character is uppercase,lowercase,digit or special symbol by using python built-in functions
n = input()
if n.isupper():
    print(f"{n} is a uppercase character")
elif n.islower():
    print(f"{n} is a lowercase character")
elif n.isdigit():
    print(f"{n} is a digit")
else:
    print(f"{n} is a special symbol")
```
