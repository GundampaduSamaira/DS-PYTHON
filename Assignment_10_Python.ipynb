#### 1. Write a Python program to find words which are greater than given length k?

```python
def check_word_len_str(strr,k):
    l = strr.split()
    res = ''
    for i in l:
        if len(i)>k:
            res += i+','
    return res
s = input("Enter string: ")
k = int(input("Enter k value: "))
r = check_word_len_str(s,k)
print(f"Words greater than the given length {k} are: ",r)
```

#### 2. Write a Python program for removing i-th character from a string?

```python
def remove_ith_char(strr,i):
    res = ''
    for j in range(len(strr)):
        if j == i:
            continue
        res += strr[j]
    return res
s = input()
i = int(input())
res = remove_ith_char(s,i)
print(res)     
```

#### 3. Write a Python program to split and join a string?

```python
def split_and_join_str(strr):
    s_split = strr.split()
    s_join = " ".join(strr.split())
    return f"splitted string is: {s_split} \n joined string is: {s_join}"

i = input()
res = split_and_join_str(i)
print(res)
```

#### 4. Write a Python program to check if a given string is binary string or not?

```python
def check_if_str_is_binary(strr):
    for i in strr:
        if i == '0' or i == '1':
            continue
        else:
            return False
    return True

i = input()
if check_if_str_is_binary(i):
    print(f"{i} is binary string")
else:
    print(f"{i} is not a binary string")
```

#### 5. Write a Python program to find uncommon words from two Strings?

```python
def uncommon_words(str1,str2):
    s1 = set(str1.split())
    s2 = set(str2.split())
    res = s1.union(s2).difference(s1.intersection(s2))
    return res
s1 = input()
s2 = input()
res = uncommon_words(s1,s2)
print(res)           
```
