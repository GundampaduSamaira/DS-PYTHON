#### 1. Create a function that squares every digit of a number.  
Examples:  
square_digits(9119) --> 811181  
square_digits(2483) --> 416649  
square_digits(3212) --> 9414

```python
def square_digits(num):
    s = str(num)
    res = ""
    for i in s:
        temp = int(i)**2
        res += str(temp)
    return res
    
i = int(input())
r = square_digits(i)
print(r)
```

#### 2. Create a function that returns the mean of all digits 
Examples:  
mean(42) --> 3  
mean(12345) --> 3  
mean(666) --> 6  
Notes:  
1.The mean of all digits is the sum of digits / how many digits there are (e.g. mean of digits in 512 is 3)  
2.The mean will always be an integer.

```python
def mean(num):
    s = str(num)
    n = len(s)
    summ = 0
    for i in s:
        summ += int(i)
    mean = summ//n
    return mean
    
i = int(input())
r = mean(i)
print(r)
```

#### 3. Create a function that sorts a list and removes all duplicate items from it.  
Examples:   
setify([1,3,3,5,5]) --> [1,3,5]  
setify([4,4,4,4]) --> [4]  
setify([5,7,8,9,10,15]) --> [5,7,8,9,10,15]  
setify([3,3,3,2,1]) --> [1,2,3]  

```python
def setify(lst):
    lst.sort()
    req_list = []
    for i in lst:
        if i not in req_list:
            req_list.append(i)
    return req_list

i = list(map(int,input().split()))
r = setify(i)
print(r)
```

#### 4. Write a Python Program to Add Two Matrices?

```python
def add_Matrices(a,b):
    print(f"Inputs: {a}, {b}")
    res_matrix = []
    if len(a) == len(b):
        for i in range(len(a)):
            if len(a[i]) == len(b[i]):
                res_matrix.append([])
                for j in range(len(a[i])):
                    res_matrix[i].append(a[i][j]+b[i][j])
            else:
                print("Both matrices must contain same no of rows and columns")
    else:
        print("Both matrices must contain same no of rows and columns")
    print(f"Output: {res_matrix}")
add_Matrices([[1,0,0],[0,1,0],[0,0,1]],[[1,0,0],[0,1,0],[0,0,1]])
add_Matrices([[1,2,3],[4,5,6]],[[9,8,7],[6,5,4]])               
```

#### 5. Write a Python Program to Multiply Two Matrices?

```python
def multiply_matrices(a,b):
    res_matrix = []
    if len(a[0]) == len(b):
        for i in range(len(a[0])):
            res_matrix.append([0 for i in range(len(b[0]))])
        for j in range(len(a)):
            for k in range(len(b[0])):
                for p in range(len(b)):
                    res_matrix[j][k] += a[j][p] * b[p][k]
        print(res_matrix)
    else:
        print("Matrix Multiplication is Not Possible")

multiply_matrices([[1,2,3],[4,5,6],[7,8,9]], [[1,4,7],[2,5,8],[3,6,9]])
```
