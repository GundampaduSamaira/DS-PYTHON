#### 1.Write a Python Program to Find the Factorial of a Number?

```python
def factorial(n):
    fact = 1
    for i in range(1,n+1):
        fact *= i
    return fact
n = int(input())
res = factorial(n)
print(res)
```

#### 2. Write a Python Program to Display the multiplication Table?

```python
def table(n):
    for i in range(1,11):
        print(f"{n} X {i} = {n*i}")
n = int(input())
table(n)
```

#### 3. Write a Python Program to Print the Fibonacci sequence?

```python
def fibonacci(n):
    s1 = 0 
    s2 = 1
    s3 = s1 + s2
    print(f"Fibnacci Squencece: {s1} {s2} {s3}",end=" ")
    for i in range(n):
        s1 = s2
        s2 = s3
        s3 = s1 + s2
        print(s3,end = " ")
n = int(input())
fibonacci(n)
```

#### 4. Write a Python Program to Check Armstrong Number?

```python
def isArmstrong(num):
    n = len(num)
    summ = 0
    for i in num:
        summ += pow(int(i),n)
    if summ == int(num):
        return True
    else:
        return False
num = input()
if isArmstrong(num):
    print(f"{num} is an Armstrong Number")
else:
    print(f"{num} is not an Armstrong Number")
```

#### 5. Write a Python Program to Find Armstrong Number in an Interval?

```python
def isArmstrong(num):
    strr = str(num)
    n = len(strr)
    summ = 0
    for i in strr:
        summ += pow(int(i),n)
    if summ == num:
        return True
    else:
        return False
start = int(input("Enter Start Interval: "))
end = int(input("Enter End Interval: "))
print(f"Armstrong Numbers from {start} to {end} are: ")
for i in range(start,end+1):
    if isArmstrong(i):
        print(i,end = " ")
    else:
        continue       
```

#### 6. Write a Python Program to Find the Sum of Natural Numbers?

```python
def sum_n_naturalNums(n):
    summ = (n*(n+1))//2
    return summ

n = int(input())
res = sum_n_naturalNums(n)
print(res)
```
